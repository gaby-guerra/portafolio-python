# Gráfica interactiva por mes

```python
# Generamos la columna anio_mes
data['fecha'] = pd.to_datetime(data['fecha'])
data['anio_mes'] = data['fecha'].dt.to_period('M')



# Agrupamos los datos
#agrupado = data.groupby(['tienda', 'anio_mes']).precio.sum().to_frame()

# Reset de los indices
#agrupado.reset_index(inplace=True)

# Se puede agrupar sin el frame
agrupado = (
    data
    .groupby(['tienda', 'anio_mes'], as_index=False)['precio']
    .sum()
)

#ordenamos los periodos y las tiendas
agrupado = agrupado.sort_values('anio_mes')
orden_tiendas = sorted(agrupado['tienda'].unique())


# Creando una columna con el valor acumulado
agrupado['acumulado'] = agrupado.groupby('tienda').precio.cumsum()

#Generar gráfico
graf = px.bar(agrupado,
              x = 'acumulado',
              y = 'tienda',
              color = 'tienda',
              text_auto = True,
              range_x = [0,160000],
              animation_frame = 'anio_mes',
              category_orders = {'tienda': orden_tiendas}
              )
graf.show()
```

# Ranking de ventras por mes

```python
import pandas as pd
import plotly.express as px

# 1) Asegurar fecha + anio_mes
data['fecha'] = pd.to_datetime(data['fecha'])
data['anio_mes'] = data['fecha'].dt.to_period('M').astype(str)

# 2) Ventas del mes por tienda
mensual = (
    data.groupby(['tienda', 'anio_mes'], as_index=False)['precio']
        .sum()
        .sort_values('anio_mes')
)

# 3) Gráfica ranking (bar race)
graf = px.bar(
    mensual,
    x='precio',
    y='tienda',
    color='tienda',
    text='precio',                 # usaremos formato custom
    animation_frame='anio_mes',
    range_x=[0, mensual['precio'].max() * 1.15],
    labels={
        'anio': 'Año',
        'precio': 'Ventas (USD)',
        'tienda': 'Tiendas'
    }
)

# 4) Formato: USD + separador de miles (ej: $12,345)
graf.update_traces(texttemplate='$%{x:,.0f}', textposition='outside')

# 5) Eje X con separador de miles también
graf.update_xaxes(tickformat='$,.0f')

# 6) Ranking dinámico: orden por valor del frame
# (si el #1 te queda abajo, cambia a 'total descending')
graf.update_layout(yaxis={'categoryorder': 'total ascending'})

graf.show()
```

# Ranking por año con pausas

```python
import pandas as pd
import plotly.express as px

# 1) Tomar de la fecha el anio
data['anio'] = pd.to_datetime(data['fecha']).dt.year

# 2) Ventas anuales por tienda
anual = (
    data.groupby(['tienda', 'anio'], as_index=False)['precio']
        .sum()
)

# 3) Gráfica ranking anual
graf = px.bar(
    anual,
    x='precio',
    y='tienda',
    color='tienda',
    text='precio',
    animation_frame='anio',
    range_x=[0, anual['precio'].max() * 1.15],
    labels={
        'anio': 'Año',
        'precio': 'Ventas (USD)',
        'tienda': 'Tienda'
    }
)

# 4) Formato: USD + separador de miles (ej: $12,345)
graf.update_traces(texttemplate='$%{x:,.0f}', textposition='outside')

# 5) Eje X con separador de miles también
graf.update_xaxes(tickformat='$,.0f')

# 6) Ranking dinámico: orden por valor del frame
graf.update_layout(yaxis={'categoryorder': 'total ascending'})

# Pausas anuales
graf.update_layout(
    updatemenus=[{
        "buttons": [{
            "args": [
                None,
                {
                    "frame": {"duration": 5000, "redraw": True},
                    "transition": {"duration": 500},
                    "fromcurrent": True
                }
            ],
            "method": "animate"
        }]
    }]
)

#título dinámico
graf.update_layout(
    title={
        'text': "Ventas por Tienda",
        'x': 0.5,
        'xanchor': 'center'
    }
)

for frame in graf.frames:
    frame.layout = {
        "title_text": f"Ventas por Tienda — Año {frame.name}"
    }

graf.show()
```
